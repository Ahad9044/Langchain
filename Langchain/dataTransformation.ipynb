```python
# Recursive Text Splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document

docs = [Document(page_content="This is a long text by Ahad " * 50)]

splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
final_documents = splitter.split_documents(docs)

print(final_documents)
```

```python
# HTML text splitter
 
from langchain_text_splitters import HTMLHeaderTextSplitter

url ="https://www.wikihow.com/Hold-Your-Poop"

headers_to_split= [("h1" , "Header 1"), 
  ("h2" , "Header2") , ("h3" , "Header3") , ("h4", "Header4") ]

splitter_html= HTMLHeaderTextSplitter(headers_to_split)
html_splitted = splitter_html.split_text_from_url(url)
html_splitted
```
