```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter

# Step 1- Text Loading

# Saves the path of file or create a file object
loader = TextLoader("speech.txt")

# load the file
documents= loader.load()
```

```python
# Step 2 - Text Splitting

# Split the long paragraph into the given chunk size 
text_splitter= CharacterTextSplitter(chunk_size=1000, chunk_overlap=30)

# preapres a new document with the specified chunk size
docs= text_splitter.split_documents(documents) 
```

```python
# Step 3 - Create the Embedding

# Initializes an embedding model (runs locally via Ollama)
embeddings = OllamaEmbeddings(model="nomic-embed-text")

# Converts each chunk (docs) into vectors and Stores them in a FAISS vector database
db= FAISS.from_documents(docs , embeddings)
```

```python
# Step 4- Querying 

query= "What is the context of the whole document?"

# The query is converted into an embedding (vector) and  That vector is compared with all stored vectors in the FAISS DB
# It finds the most similar chunks of text and returns them as a list of Document objects

docs= db.similarity_search(query)
docs[0].page_content
```
